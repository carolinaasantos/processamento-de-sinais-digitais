# Broadcasting com arrays NumPy

O termo broadcasting (ou transmissão) se refere a como o NumPy trata arrays com dimensões diferentes durante operações aritméticas, o que resulta em certas restrições. O array menor é expandido (broadcast) para que tenha formas compatíveis com o array maior.

O __broadcasting__ oferece uma __maneira de vetorizar operações em arrays__, de modo que o __loop ocorra em C em vez de Python__, já que sabemos que o NumPy é implementado em C. Ele faz isso sem criar cópias desnecessárias de dados, resultando em __implementações eficientes de algoritmos__. Há casos em que o broadcasting é uma má ideia, pois pode levar a um uso ineficiente da memória, o que torna o cálculo mais lento.


__Regras de Broadcasting:__

A transmissão (broadcasting) de dois arrays juntos segue estas regras:

1. Se as matrizes não tiverem a mesma classificação, prefixe a forma da matriz de classificação inferior com 1s até que ambas as formas tenham o mesmo comprimento.
2. As duas matrizes são compatíveis em uma dimensão se tiverem o mesmo tamanho na dimensão ou se uma das matrizes tiver tamanho 1 nessa dimensão.
3. As matrizes podem ser transmitidas juntas se forem compatíveis com todas as dimensões.
4. Após a transmissão, cada array se comporta como se tivesse um formato igual ao máximo em termos de elementos dos dois arrays de entrada.
5. Em qualquer dimensão em que uma matriz tenha tamanho 1 e a outra matriz maior que 1, a primeira matriz se comporta como se tivesse sido copiada ao longo dessa dimensão.

## Broadcasting de array unidimensional

```python
a = np.array([5, 10, 15]) # 1x3 Dimension array
print(a)
b = 3
print(b)

# Broadcasting happened because of
# miss match in array Dimension.
c = a + b
d = a * b
print(c)
print(d)
```

## Broadcasting de array bidimensional

```python
A = np.array([[11, 22, 33], [10, 20, 30]])
print(A)

b = 4
print(b)

C = A + b
print(C)

D = A * 2
print(D)
D = np.reshape(D, (6, 1))
print(D)
```

## Broadscating na plotagem de uma função bidimensional:

A transmissão (broadcasting) também é frequentemente usada na exibição de imagens baseadas em funções bidimensionais. Se quisermos definir uma função $z=f(x, y)$.

```python
import numpy as np
import matplotlib.pyplot as plt

# Computes x and y coordinates for
# points on sine and cosine curves
x = np.arange(0, 3 * np.pi, 0.1)
y_sin = np.sin(x)
y_cos = np.cos(x)

# Plot the points using matplotlib
plt.plot(x, y_sin)
plt.plot(x, y_cos)
plt.xlabel('x axis label')
plt.ylabel('y axis label')
plt.title('Sine and Cosine')
plt.legend(['Sine', 'Cosine'])

plt.show()
```

# Estude com atenção o exemplo a seguir

O objetivo deste exemplo é gerar observações (amostras) ruidosas a partir de uma função objetivo e criar uma figura comparando-as.

```python
# Importando as bibliotecas necessárias.
import numpy as np
import matplotlib.pyplot as plt

# Reseta o gerador de sequências pseudo-aleatórias.
np.random.seed(42)

# Define o número de amostras.
N = 1000

# Vetor coluna com dimensão Nx1, com valores linearmente espaçados entre -1 e 1.
x = np.linspace(-1, 1, N).reshape(N,1)

# Vetor ruído com dimensão Nx1 e variância igual a 0.1.
w = np.sqrt(0.1)*np.random.randn(N,1)

# Função original.
y = -1 + 2*x**2

# Versão ruidosa de y.
y_noisy = y + w

# Plotagem
plt.plot(x, y_noisy, '.b', label='Função ruidosa')
plt.plot(x, y, 'k', label='Função original', linewidth=4)
plt.xlabel('x', fontsize=14)
plt.ylabel('y', fontsize=14)
plt.legend()
plt.grid()

# salva figura em arquivo
plt.savefig('figura_2D.png')

# Mostra a figura.
plt.show()
```

# __Exercício 8:__

Usando o conhecimento adquirido nos exemplos anteriores resolva o seguinte exercício.

Neste exercício você irá plotar um gráfico 2D. Este tipo de gráfico é comumente
utilizado para se analisar os dados de entrada e saída de um modelo de aprendizado de
máquina. Crie um vetor coluna, $y$, com $M = 1000$ elementos, onde $y$ é dado pela
seguinte equação

$$y=1.2 + 2.3*x + 10*w$$

em que $x$ é um vetor coluna com $M$ elementos retirados de uma __distribuição aleatória
uniforme__ com valores no intervalo em $[0, 1)$ e $w$ é um vetor coluna com $M$ elementos
retirados de uma __distribuição aleatória Gaussiana normal__, i.e., com média 0 e variância
unitária.

- Plote um gráfico com os vetores $x$ e $y$ sendo os eixos $x$ e $y$, respectivamente. Cada par de valores $(x,y)$ deve ser mostrado no gráfico como sendo um ponto.

__Dicas:__
- Use o módulo random da biblioteca numpy para gerar números aleatórios https://docs.scipy.org/doc/numpy-1.14.0/reference/routines.random.html
- Use a biblioteca matplotlib para plotar gráficos
https://matplotlib.org/3.1.3/gallery/lines_bars_and_markers/simple_plot.html

```python
# Inserir o seu código abaixo:
# Importando as bibliotecas necessárias.
import numpy as np
import matplotlib.pyplot as plt

# Reseta o gerador de sequências pseudo-aleatórias.
np.random.seed(24)

M = 1000
x = np.random.uniform(0, 1, M).reshape(M,1)
w = np.random.normal(0, 1, M).reshape(M,1)
y = 1.2 + 2.3 * x + 10 * w
y_normal = 1.2 + 2.3 * x

# Plotagem
plt.scatter(x, y, c='b', marker='o', s=10, label='Ruídos')
plt.plot(x, y_normal, 'r', label='Y sem ruídos', linewidth=4)
plt.xlabel('x', fontsize=14)
plt.ylabel('y', fontsize=14)
plt.legend()
plt.grid()

# Mostra a figura.
plt.show()
```

# Como criar e plotar histogramas - sinal 1D

```python

data = np.random.randn(1000000)

#
# Note o uso da função subplot para plotar o histograma - probability density function (pdf)
#
plt.subplot(1, 2, 1)
plt.title('PDF')
plt.hist(data, bins=100, density=True, color='b')

# Cumulative density function (CDF) empirica
plt.subplot(1, 2, 2)
plt.title('CDF')
plt.hist(data, bins=100, density=True, color='g', cumulative=True)

plt.savefig('histogram.png') # salva figura em arquivo
```
